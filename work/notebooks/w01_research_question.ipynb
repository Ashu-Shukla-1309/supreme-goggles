The Question: Which currently visible content pages are at the highest risk of traffic decay, and therefore should be prioritized for a content refresh?

Unit of Analysis: A single page (content item) at a specific point in time.

The Decision & Action: An SEO manager looks at the ranked list and assigns the top pages to a writer to update, expand, or consolidate the content.

Cost of a wrong call:

-False Positive (Flagging a healthy page): Wasted editorial budget and writer time trying to fix something that isn't broken.

-False Negative (Missing a decaying page): The page continues to lose rankings, resulting in lost traffic, fewer conversions, and lost revenue.

Why ML helps: Hand-written rules (like "stale + high impressions") are rigid and treat all decaying pages identically. ## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess
import pandas as pd
import numpy as np

# 1. Setup: Clone repo if in Colab and set the correct working directory
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    # Move into the repo root if not already there
    if os.path.basename(os.getcwd()) != REPO_DIR:
        os.chdir(REPO_DIR)

# 2. Load the starter data (now using the path relative to the repo root)
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 3. Base Rate of Underperformance (Pages losing traffic)
declining_pct = (df["trend_direction"].str.lower() == "down").mean() * 100

# 4. Finding the "Opportunity Gap" (High historical potential, currently failing)
opportunity_pages = df[(df["avg_position"] <= 10) & (df["ctr"] < 0.01)]

# 5. Size of the missed opportunity
missed_impressions = opportunity_pages["impressions_90d"].sum()

print(f"1. Decay Rate: {declining_pct:.1f}% of all pages are in a downward trend, providing a massive target for our expected-value model.")
print(f"2. Actionable Gap: There are {len(opportunity_pages)} pages ranking on page 1 (position <= 10) but severely underperforming with a CTR under 1%.")
print(f"3. Scale: These specific underperforming pages are sitting on {missed_impressions:,.0f} recent impressions, representing immediate, recoverable ROI for an Action Engine.")
```

What this proves: The problem has massive scale. Millions of impressions are flowing through pages that rank well but fail to capture clicks. An expected-value model can mathematically quantify this "missed opportunity" and rank these pages by recoverable traffic, rather than just historical volume.
